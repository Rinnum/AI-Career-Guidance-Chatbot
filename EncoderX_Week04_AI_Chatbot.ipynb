````python
import time

def chatbot_response(user_message, conversation_history):

    system_prompt = """
You are an AI Career Guidance Assistant.

Your purpose is to help university students and fresh graduates
with technology careers and learning.

You can help with:
- Career guidance
- Programming
- Artificial Intelligence
- Machine Learning
- Data Science
- Cybersecurity
- Software Development
- Technical interviews
- Resume and CV improvement
- Learning roadmaps
- Student project ideas

Give clear, practical and beginner-friendly answers.

If the user asks a question unrelated to career, education,
technology or learning, politely explain that you are mainly
designed for career and technology guidance.

Do not make up facts.
"""

    history_text = ""

    for message in conversation_history:
        history_text += f"{message['role']}: {message['content']}\n"

    prompt = f"""
{system_prompt}

Previous conversation:
{history_text}

User:
{user_message}

Assistant:
"""

    # Try the main model
    models_to_try = [
        "gemini-3.5-flash",
        "gemini-3.5-flash-lite"
    ]

    last_error = None

    for model_name in models_to_try:

        for attempt in range(3):

            try:
                response = client.models.generate_content(
                    model=model_name,
                    contents=prompt
                )

                return response.text

            except Exception as e:

                last_error = e

                print(
                    f"Attempt {attempt + 1} failed "
                    f"with {model_name}. Retrying..."
                )

                time.sleep(3)

    return f"Sorry, the AI service is temporarily unavailable. Please try again later.\n\nError: {last_error}"
````

````python
user_message = "I want to become an AI engineer."

answer = chatbot_response(
    user_message,
    conversation_history
)

conversation_history.append({
    "role": "user",
    "content": user_message
})

conversation_history.append({
    "role": "assistant",
    "content": answer
})

print("Assistant:")
print(answer)
````

````python
%%writefile app.py

import streamlit as st
from google import genai

# -----------------------------
# Gemini API Configuration
# -----------------------------

API_KEY = st.secrets["GEMINI_API_KEY"]

client = genai.Client(api_key=API_KEY)

# -----------------------------
# Page Configuration
# -----------------------------

st.set_page_config(
    page_title="AI Career Guidance Chatbot",
    page_icon="🤖",
    layout="centered"
)

st.title("🤖 AI Career Guidance Chatbot")
st.write(
    "Ask questions about programming, AI, cybersecurity, "
    "software development, careers, interviews, and learning roadmaps."
)

# -----------------------------
# System Prompt
# -----------------------------

SYSTEM_PROMPT = """
You are an AI Career Guidance Assistant.

Your purpose is to help university students and fresh graduates
with technology careers and learning.

You can help with:
- Career guidance
- Programming
- Artificial Intelligence
- Machine Learning
- Data Science
- Cybersecurity
- Software Development
- Technical interviews
- Resume and CV improvement
- Learning roadmaps
- Student project ideas

Give clear, practical and beginner-friendly answers.

If the user asks something unrelated to career, education,
technology or learning, politely explain that you are mainly
designed for career and technology guidance.

Do not make up facts.
"""

# -----------------------------
# Conversation History
# -----------------------------

if "messages" not in st.session_state:
    st.session_state.messages = []

# -----------------------------
# Display Previous Messages
# -----------------------------

for message in st.session_state.messages:

    with st.chat_message(message["role"]):
        st.markdown(message["content"])

# -----------------------------
# Chat Input
# -----------------------------

user_message = st.chat_input(
    "Ask me about your technology career..."
)

if user_message:

    # Display user message
    with st.chat_message("user"):
        st.markdown(user_message)

    st.session_state.messages.append({
        "role": "user",
        "content": user_message
    })

    # Build conversation history
    history_text = ""

    for message in st.session_state.messages:
        history_text += (
            f"{message['role']}: "
            f"{message['content']}\n"
        )

    prompt = f"""
{SYSTEM_PROMPT}

Previous conversation:

{history_text}

Assistant:
"""

    # Generate response
    try:

        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=prompt
        )

        assistant_response = response.text

    except Exception as e:

        assistant_response = (
            "Sorry, the AI service is temporarily unavailable. "
            "Please try again in a moment."
        )

    # Display AI response
    with st.chat_message("assistant"):
        st.markdown(assistant_response)

    # Save response
    st.session_state.messages.append({
        "role": "assistant",
        "content": assistant_response
    })

# -----------------------------
# Clear Chat
# -----------------------------

if st.button("🗑️ New Chat"):

    st.session_state.messages = []

    st.rerun()
````

````python
%%writefile README.md

# 🤖 AI Career Guidance Chatbot

An AI-powered career guidance chatbot developed as part of the EncoderX Remote Internship – AI/ML Week 04 task.

## 📌 Project Overview

The AI Career Guidance Chatbot is designed to help university students and fresh graduates with technology-related career and learning guidance.

The chatbot can provide guidance about:

- Programming
- Artificial Intelligence
- Machine Learning
- Data Science
- Cybersecurity
- Software Development
- Technical Interviews
- Resume and CV Improvement
- Learning Roadmaps
- Student Project Ideas

## 🎯 Objective

The objective of this project is to develop an intelligent chatbot that can understand user queries and generate meaningful responses using Generative AI.

## 🛠️ Technologies Used

- Python
- Google Gemini API
- Google GenAI SDK
- Streamlit
- Google Colab
- GitHub

## 🧠 AI Integration

The chatbot uses the Google Gemini API to generate AI-powered responses.

A system prompt is used to define the chatbot's purpose and behavior. The chatbot provides beginner-friendly and practical career and technology guidance.

## 💬 Conversation History

The application maintains conversation history using Streamlit session state.

Previous user and assistant messages are included when generating new responses, allowing the chatbot to maintain conversational context.

## 🖥️ User Interface

The chatbot interface is developed using Streamlit.

Features include:

- Chat input
- User messages
- AI responses
- Conversation history
- New Chat button

## 📂 Project Structure

```text
AI-Career-Guidance-Chatbot/
│
├── app.py
├── requirements.txt
└── README.md
````
